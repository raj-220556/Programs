```python
# LINE PLOT

# THEORY:
# A Line Plot is used to visualize the relationship between two continuous variables.
# It shows how a variable changes with respect to another variable.
#
# The x-axis usually represents time, sequence, or independent variable.
# The y-axis represents the dependent variable whose values change.
#
# In a line plot:
# 1. Individual data points (x,y) are first plotted on the graph.
# 2. These points are then connected with straight line segments.
# 3. The line helps visualize trends such as increase, decrease, or stability.
#
# Example interpretation:
# If x = days and y = temperature,
# each point represents temperature on a particular day.
#
# If temperature values are:
# Day 1 → 30
# Day 2 → 32
# Day 3 → 31
#
# The graph will show:
# (1,30) → (2,32) → (3,31)
#
# The line between them helps visualize the trend of temperature changes.
#
# Line plots are commonly used for:
# - stock prices
# - population growth
# - sales trends
# - sensor readings over time

import matplotlib.pyplot as plt

days = [1,2,3,4,5]
temperature = [30,32,31,35,36]

plt.plot(days, temperature)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Temperature Change Over Days")

plt.show()
```

```python
# AREA PLOT

# THEORY:
# An Area Plot is similar to a line plot but the region under the line is filled.
#
# Purpose:
# It emphasizes the magnitude of values rather than only the trend.
#
# In an area plot:
# 1. Data points are plotted like a line plot.
# 2. The space between the line and x-axis is filled with color.
#
# This helps visualize accumulated quantities.
#
# Example:
# If daily sales are:
# Day 1 → 100
# Day 2 → 120
# Day 3 → 150
#
# The filled area visually shows total growth in sales.
#
# Area plots are useful for:
# - revenue growth
# - cumulative totals
# - resource usage over time

import matplotlib.pyplot as plt

days = [1,2,3,4,5]
sales = [100,120,150,170,200]

plt.fill_between(days, sales)

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Growth (Area Plot)")

plt.show()
```

```python
# HISTOGRAM

# THEORY:
# A histogram is used to visualize the frequency distribution of numerical data.
# Instead of plotting each value separately, the data is grouped into intervals called bins.
#
# BINS:
# A bin is a range of values.
#
# Example dataset:
# marks = [55,60,65,70,72,75,80,82,85,88,90,92,95]
#
# If bins = 5
#
# First the algorithm finds:
# minimum value = 55
# maximum value = 95
#
# Range = max - min
# Range = 95 - 55 = 40
#
# Bin width = Range / number of bins
# Bin width = 40 / 5 = 8
#
# So bins become approximately:
#
# Bin1 → 55 - 63
# Bin2 → 63 - 71
# Bin3 → 71 - 79
# Bin4 → 79 - 87
# Bin5 → 87 - 95
#
# Now the histogram counts how many values fall in each bin.
#
# Example:
# Bin1 (55–63) → 55,60 → 2 values
# Bin2 (63–71) → 65,70 → 2 values
# Bin3 (71–79) → 72,75 → 2 values
# Bin4 (79–87) → 80,82,85 → 3 values
# Bin5 (87–95) → 88,90,92,95 → 4 values
#
# Each bin becomes a bar whose height represents frequency.

import matplotlib.pyplot as plt

marks = [55,60,65,70,72,75,80,82,85,88,90,92,95]

plt.hist(marks, bins=5)

plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.title("Marks Distribution")

plt.show()
```

```python
# BAR CHART

# THEORY:
# A bar chart represents categorical data using rectangular bars.
#
# Each bar represents one category.
# The height (or length) of the bar represents the value associated with that category.
#
# Example:
# Subjects = Math, Physics, Chemistry
# Marks = 85, 78, 90
#
# The graph will show three bars:
#
# Math → height 85
# Physics → height 78
# Chemistry → height 90
#
# Bars are separated because categories are discrete (not continuous).
#
# Bar charts are used for:
# - comparing categories
# - survey results
# - product sales comparison

import matplotlib.pyplot as plt

subjects = ["Math","Physics","Chemistry","Biology"]
marks = [85,78,90,75]

plt.bar(subjects, marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks in Different Subjects")

plt.show()
```

```python
# PIE CHART

# THEORY:
# A pie chart represents parts of a whole.
#
# The entire circle represents 100% of the data.
#
# Each slice represents the proportion of one category.
#
# Example dataset:
# Python = 40
# Java = 25
# C++ = 20
# JavaScript = 15
#
# Total = 100
#
# Percentage calculation:
#
# Python → 40 / 100 = 40%
# Java → 25 / 100 = 25%
# C++ → 20 / 100 = 20%
# JavaScript → 15 / 100 = 15%
#
# Each percentage determines the angle of the slice.
#
# Angle formula:
# angle = percentage × 360°
#
# Example:
# Python slice = 0.40 × 360 = 144°

import matplotlib.pyplot as plt

labels = ["Python","Java","C++","JavaScript"]
students = [40,25,20,15]

plt.pie(students, labels=labels, autopct='%1.1f%%')

plt.title("Programming Language Preference")

plt.show()
```

```python
# BOX PLOT

# THEORY:
# A Box Plot (also called Box-and-Whisker Plot) is used to visualize the statistical
# distribution of a dataset. It summarizes the dataset using five important values
# called the Five Number Summary.

# FIVE NUMBER SUMMARY:
# 1. Minimum value
# 2. First Quartile (Q1)
# 3. Median (Q2)
# 4. Third Quartile (Q3)
# 5. Maximum value

# Example dataset
data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 100]

# STEP 1 — Sort the data
# Sorted data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 100]

# STEP 2 — Find Median (Q2)
# Number of elements = 10 (even)
# Median = average of 5th and 6th values
# Q2 = (20 + 22) / 2 = 21

# STEP 3 — Find Q1 (First Quartile)
# Lower half = [10, 12, 15, 18, 20]
# Median of lower half = 15
# Q1 = 15

# STEP 4 — Find Q3 (Third Quartile)
# Upper half = [22, 25, 30, 35, 100]
# Median of upper half = 30
# Q3 = 30

# So we get:
# Minimum = 10
# Q1 = 15
# Median (Q2) = 21
# Q3 = 30
# Maximum = 100

# INTERQUARTILE RANGE (IQR)
# IQR = Q3 - Q1
# IQR = 30 - 15 = 15

# OUTLIER DETECTION
# Outliers are extreme values that lie far away from the rest of the data.

# Lower Bound = Q1 - 1.5 * IQR
# Lower Bound = 15 - (1.5 × 15) = 15 - 22.5 = -7.5

# Upper Bound = Q3 + 1.5 * IQR
# Upper Bound = 30 + (1.5 × 15) = 30 + 22.5 = 52.5

# Any value < -7.5 or > 52.5 is an OUTLIER.

# In this dataset:
# 100 > 52.5
# So 100 is an OUTLIER.

# BOX PLOT STRUCTURE:
# The box represents values from Q1 to Q3.
# The line inside the box represents the Median (Q2).
# The whiskers extend from the box to the minimum and maximum values
# excluding outliers.
# Outliers are shown as individual points.

import matplotlib.pyplot as plt

plt.boxplot(data)

plt.title("Box Plot Example Showing Quartiles and Outlier")

plt.show()
```

```python
# SCATTER PLOT

# THEORY:
# A scatter plot displays the relationship between two numerical variables.
#
# Each point on the graph represents one observation.
#
# Example:
# Study hours → x-axis
# Marks → y-axis
#
# If a student studies 3 hours and scores 55 marks,
# the point plotted is (3,55).
#
# The pattern of points indicates correlation:
#
# Positive correlation:
# points move upward (both variables increase)
#
# Negative correlation:
# points move downward (one increases while other decreases)
#
# No correlation:
# points appear randomly scattered

import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6]
marks = [40,50,55,65,70,80]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
# BUBBLE PLOT

# THEORY:
# A bubble plot is an extension of the scatter plot.
#
# It displays three variables simultaneously:
#
# x coordinate → first variable
# y coordinate → second variable
# bubble size → third variable
#
# Example:
#
# x = study hours
# y = marks
# size = number of assignments completed
#
# Larger bubble → larger value of third variable
#
# Bubble size is controlled using parameter s.
#
# If s = [100,200,300]
# it means bubble sizes increase for each point.

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,30,25,15]

sizes = [100,300,500,700,900]

plt.scatter(x, y, s=sizes)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Bubble Plot Example")

plt.show()
```
