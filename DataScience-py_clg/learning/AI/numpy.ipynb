```python
import numpy as np

# Packet counts per second
traffic_data = [45, 48, 52, 49, 1200, 51, 47, 46, 1150, 48, 50, 52] # ... etc

# 1. Convert to array
data = np.array(traffic_data)

# 2. Calculate stats
mean_val = np.mean(data)
std_val = np.std(data)

# 3. Vectorized Filtering (The "Magic")
outliers = data[data > (mean_val + 2 * std_val)]

print(f"Mean: {mean_val}")
print(f"Spikes Detected: {outliers}")

# Original Data:"[45, 1200, 51, 1150]"
# The Logic:data > 100
# The Mask:"[False, True, False, True]"
```
