````python
import re

# Pre-compile the pattern for speed
ip_pattern = re.compile(r"([0-9]{1,3}\.){3}[0-9]{1,3}")

def stream_logs(filepath):
    result = []
    with open(filepath, 'r') as file:
        for line in file:
            match = ip_pattern.search(line)
            if match:
                result.append(match.group(0))
    return result
str1 = stream_logs("/var/log/syslog")
print(len(str1))
````

````python
import re
log = "ERROR 2026-02-12: Unauthorized access attempt from 10.0.0.5"

# Search for the word 'Unauthorized' anywhere in the line
match = re.search(r"Unauthorized", log)
if match:
    print(f"Alert Triggered! Found at index: {match.start()}")
````

````python
log = "192.168.1.1 - GET /admin"

# This works because the IP is at the start
if re.match(r"\d{1,3}\.", log):
    print("Valid log format.")

# This fails because 'GET' is not at the start
if not re.match(r"GET", log):
    print("Pattern not at start of string.")
````

````python
email_body = "Contact us at support@malware.com or admin@phish.net"
emails = re.findall(r"[\w\.-]+@[\w\.-]+", email_body)

print(emails) # Output: ['support@malware.com', 'admin@phish.net']
````

````python
raw_data = "Port:80, Port:443, Port:8080"
matches = re.finditer(r"Port:(\d+)", raw_data)

for m in matches:
    print(f"Found Port: {m.group(1)} at position {m.start()}")
````
