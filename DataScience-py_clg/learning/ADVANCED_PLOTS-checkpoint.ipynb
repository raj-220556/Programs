```python
# WAFFLE CHART

# THEORY:
# A Waffle Chart is used to represent categorical data using small square blocks.
# It is similar to a pie chart but easier to read visually.

# The chart is built as a grid of squares (usually 10 × 10 = 100 squares).
# Each square represents a portion of the total percentage.

# Example:
# Suppose we have programming language popularity:
# Python = 40
# Java = 30
# C++ = 20
# JavaScript = 10

# Total = 100

# If we create a 10x10 grid (100 blocks):
# Python → 40 blocks
# Java → 30 blocks
# C++ → 20 blocks
# JavaScript → 10 blocks

# Each block represents 1% of the data.

# This type of visualization makes proportions easier to understand
# compared to pie charts.

from pywaffle import Waffle
import matplotlib.pyplot as plt

data = {'Python':40,'Java':30,'C++':20,'JavaScript':10}

fig = plt.figure(
    FigureClass=Waffle,
    rows=10,
    values=data,
    title={'label':'Programming Language Popularity'}
)

plt.show()
```

```python
# WORD CLOUD

# THEORY:
# A Word Cloud is used to visualize text data.
# Words appear in different sizes depending on their frequency.

# The more frequently a word appears in the text,
# the larger it will appear in the visualization.

# Example text:
# "data science python data analysis machine learning python data"

# Word frequency:
# data → 3 times
# python → 2 times
# science → 1 time

# In the word cloud:
# "data" will appear the largest
# "python" medium size
# "science" smaller

# Word clouds are used in:
# - social media analysis
# - customer reviews analysis
# - text mining
# - NLP visualization

from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = """
data science python data analysis machine learning
python visualization data science statistics python
"""

wc = WordCloud(width=800,height=400,background_color='white').generate(text)

plt.imshow(wc)
plt.axis("off")
plt.title("Word Cloud Example")

plt.show()
```

```python
# SEABORN PLOT

# THEORY:
# Seaborn is a statistical data visualization library built on top of Matplotlib.
# It provides more attractive and informative graphics.

# Seaborn automatically:
# - improves plot styling
# - handles complex statistical relationships
# - works well with pandas dataframes

# In this example we create a scatter plot using seaborn.

# Each point represents one observation.
# The plot shows the relationship between study hours and marks.

import seaborn as sns
import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6,7]
marks = [40,50,55,65,70,80,90]

sns.scatterplot(x=study_hours,y=marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Seaborn Scatter Plot")

plt.show()
```

```python
# REGRESSION PLOT

# THEORY:
# A regression plot shows the relationship between two variables
# along with a regression line (best fit line).

# The regression line predicts the expected value of Y
# for each value of X.

# Example:
# X = study hours
# Y = marks

# If the regression line slopes upward,
# it indicates positive correlation.

# Positive correlation:
# As study hours increase, marks also increase.

# Negative correlation:
# As one variable increases, the other decreases.

# The regression line is calculated using statistical methods
# like Least Squares Method.

# Regression plots are commonly used in:
# - machine learning
# - predictive modeling
# - data analysis

import seaborn as sns
import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6,7]
marks = [40,50,55,65,70,80,90]

sns.regplot(x=study_hours,y=marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Regression Plot")

plt.show()
```
