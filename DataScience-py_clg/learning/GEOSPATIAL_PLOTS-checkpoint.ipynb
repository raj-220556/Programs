```python
# BASIC MAP

# THEORY:
# Geospatial visualization means representing geographical data
# (location-based data) on maps.

# These visualizations help analyze spatial patterns such as:
# population distribution
# weather patterns
# traffic data
# store locations

# Folium is a Python library used to create interactive maps.
# It is based on Leaflet.js (a JavaScript mapping library).

# The basic map is created using coordinates:
# Latitude → vertical position
# Longitude → horizontal position

# Example coordinates:
# Vijayawada
# Latitude  = 16.5062
# Longitude = 80.6480

import folium

# Create map centered at Vijayawada
map = folium.Map(location=[16.5062, 80.6480], zoom_start=12)

map
```

```python
# MAP WITH MARKERS

# THEORY:
# Markers are used to highlight specific locations on a map.

# A marker represents a point of interest such as:
# hospitals
# restaurants
# tourist places
# schools

# Each marker can contain:
# popup text
# icon
# tooltip

# When the user clicks the marker,
# the popup information appears.

import folium

map = folium.Map(location=[16.5062, 80.6480], zoom_start=12)

# Add marker
folium.Marker(
    location=[16.5062, 80.6480],
    popup="Vijayawada City",
    tooltip="Click here"
).add_to(map)

map
```

```python
# MULTIPLE MARKERS

# THEORY:
# Multiple markers are used when we want to show several locations
# on the same map.

# Example:
# showing multiple cities or multiple stores.

# Each marker represents one location.

import folium

map = folium.Map(location=[16.5062, 80.6480], zoom_start=6)

cities = [
    ["Vijayawada",16.5062,80.6480],
    ["Hyderabad",17.3850,78.4867],
    ["Chennai",13.0827,80.2707],
    ["Bangalore",12.9716,77.5946]
]

for city in cities:
    folium.Marker(
        location=[city[1],city[2]],
        popup=city[0]
    ).add_to(map)

map
```

```python
# CHOROPLETH MAP

# THEORY:
# A choropleth map represents data using color intensity
# across geographic regions.

# Different colors represent different data values.

# Example uses:
# population by state
# COVID cases by country
# literacy rate by region
# rainfall distribution

# Darker color → higher value
# Lighter color → lower value

# Choropleth maps require:
# 1 GeoJSON file (map boundaries)
# 2 Dataset containing values for regions

import folium

map = folium.Map(location=[20.5937,78.9629], zoom_start=5)

# Example data (dummy values)
data = {
    "Andhra Pradesh":50,
    "Telangana":70,
    "Tamil Nadu":65,
    "Karnataka":60
}

# Choropleth example structure
# (Actual implementation requires GeoJSON boundaries)

folium.Choropleth(
    geo_data="india_states.geojson",
    data=data,
    columns=["State","Value"],
    key_on="feature.properties.state"
).add_to(map)

map
```

```python
# CHOROPLETH MAP

# THEORY:
# A choropleth map represents statistical data using colors across
# geographical regions.

# Each region is filled with a color based on the value associated
# with that region.

# Darker color → higher value
# Lighter color → lower value

# It is commonly used for:
# population density
# literacy rate
# rainfall
# election results

# Choropleth maps require:
# 1. Geographic boundary data (GeoJSON)
# 2. Dataset containing values for those regions

import folium

# Create base map centered on India
map = folium.Map(location=[20.5937,78.9629], zoom_start=5)

# Example GeoJSON (simple dummy regions)
geo_data = {
"type":"FeatureCollection",
"features":[
{
"type":"Feature",
"properties":{"name":"Region1"},
"geometry":{
"type":"Polygon",
"coordinates":[[[78,15],[80,15],[80,17],[78,17],[78,15]]]
}
},
{
"type":"Feature",
"properties":{"name":"Region2"},
"geometry":{
"type":"Polygon",
"coordinates":[[[80,15],[82,15],[82,17],[80,17],[80,15]]]
}
}
]
}

# Data for regions
data = {
"Region1":50,
"Region2":80
}

# Choropleth visualization
folium.Choropleth(
geo_data=geo_data,
data=data,
columns=["Region","Value"],
key_on="feature.properties.name",
fill_color="YlOrRd",
legend_name="Sample Data"
).add_to(map)

map
```
