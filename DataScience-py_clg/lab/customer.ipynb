```python
# Part A: Data Loading & Basic Understanding (5 Tasks)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load the dataset into a pandas DataFrame.
df = pd.read_csv("dataset.csv")

print(df)  # It Prints total data in the dataset

print(df.head())  # It display first 5 rows

print(df.info()) # It Prints MetaData of dataset

print(list(df.columns)) # It returns Object

print(df.shape) # Shows (rows, clumns)
```

```python
# 🟢 Part B: Data Types & Missing Values (5 Tasks)/

# Returns a Series with column names and their types
print("Columns and Datatypes of a Dataset")
print(df.dtypes)

print("Total no.of missing values in each column")
print(df.isnull().sum()) # It Sum the not NULL valued rows in column

numcol = df.select_dtypes(include = np.number).columns
objcol = df.select_dtypes(include = 'object').columns

df[numcol] = df[numcol].fillna(df[numcol].mean())
# df[objcol] = df[objcol].fillna(df[objcol].mode().iloc[0])  # It gives error if objcol contains more columns

for col in objcol:
    col_mode = df[col].mode()
    if not df[col].mode().empty:
        df[col] = df[col].fillna(col_mode[0]) # [0] for select the First most mode object
    else:
        df[col] = df[col].fillna("UNKOWN")
    

print("Total no.of Missing values in each column after handling")
print(df.isnull().sum())

```

```python
# 🟢 Part C: Descriptive Statistics

print("Mean of all Numerical Columns")
print(df[numcol].mean())

print("Median of all Numerical Columns")
print(df[numcol].median())
print("Standard Deviation of all Numerical Columns")
print(df[numcol].std())
print("Aggregation of all Numerical Columns") # Minimum & Maximum of all Columns
print(df[numcol].agg(['min', 'max']))

print("Summary Description of all Numerical Columns")
print(df[numcol].describe())

```

```python
# 🟢 Part D: Basic Analysis & Visualization

plt.hist(df['purchase_amount'], bins=20) # It shows count of a particular value and bins: group range of values
plt.title('Histogram of Purchase Amount') # Title of a Graph
plt.xlabel('Purchase Amount') # Title of x-axis of a Graph
plt.ylabel('Frequency')  # Title of y-axis of a Graph
plt.show() # To Display Result of a graph

# CountPlot: Shows no.of customers in each purchase frequency category
sns.countplot(x='purchase_frequency', data=df)
plt.title("Customers by purchase frequency")
plt.xticks(rotation=45)
plt.show()

# BoxPlot: Shows spread of purchase amount and detects outliers
sns.boxplot(x=df['purchase_amount'])
plt.title("Purchase amount Boxplot")
plt.show()

# ScatterPlot: Shows relationship between age and purchase
sns.scatterplot(x='age', y='purchase_amount', data=df)
plt.title("Age Vs Purchase amount")
plt.show()

# Heatmap: Shows correaltion between numerical variables
# Cell values Shows correalation strength between variables
plt.figure(figsize=(8,6))
sns.heatmap(df[numcol].corr(), annot=True, cmap='coolwarm')
plt.title("Corelation Heatmap")
plt.show()
```
