# TASK 1

1. Create a waffle chart for student grades distribution.
2. Visualize department-wise student count.
3. Modify grid size (5×20) and observe changes.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pywaffle import Waffle
Writing objects:  96% (27/28), 193.98 MiB | 547.00 KiB/s

data = {
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'MECH', 'MECH', 'CIVIL', 'CIVIL', 
                   'CSE', 'CSE', 'ECE', 'ECE', 'MECH', 'MECH', 'CIVIL', 'CIVIL'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 
               'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Year': [2023, 2023, 2023, 2023, 2023, 2023, 2023, 2023, 
             2024, 2024, 2024, 2024, 2024, 2024, 2024, 2024],
    'Count': [85, 45, 60, 40, 50, 20, 35, 15, 90, 55, 65, 45, 52, 25, 38, 18]
}
df = pd.DataFrame(data)
# 1. Create a waffle chart for student grades distribution
grade_counts = {
    'Grade A': 40, 
    'Grade B': 35, 
    'Grade C': 15, 
    'Grade D': 7, 
    'Grade F': 3
}
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,                         # Number of rows
    columns=20,                      # Task 1, Q3: Modified grid size 
    values=grade_counts,
    colors=["#2ecc71", "#3498db", "#f1c40f", "#e67e22", "#e74c3c"],
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    icons='user',                    # Optional: uses FontAwesome icons
    font_size=12,
    title={'label': 'Student Grades Distribution', 'loc': 'left'},
    figsize=(10, 5)
)
plt.show()

# 2. Visualize department-wise student count.
dept_2024 = df[df['Year'] == 2024].groupby('Department')['Count'].sum().to_dict()

fig = plt.figure(
    FigureClass=Waffle,
    rows=10,
    columns=10,
    values=dept_2024,
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    title={'label': 'Department-wise Student Count (2024)', 'loc': 'left'},
    figsize=(10, 5)
)
plt.show()

# 3. Modify grid size (5×20) and observe changes.

fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    columns=20,
    values=dept_2024,
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    title={'label': 'Department-wise Student Count (2024)', 'loc': 'left'},
    figsize=(10, 5)
)
plt.show()
```

# Lab Task 2: Text Analysis (Word Cloud)
1. Project Overview
This task involves generating a Word Cloud to visualize student feedback. The goal is to identify the most frequent terms used in the feedback to understand the overall sentiment and key themes .

2. Dataset: student_feedback.csv
The following feedback was extracted from the lab manual for analysis :

Teaching is excellent and very clear

Labs are well organized and very helpful

The faculty explains machine learning clearly

Python lab sessions are very interesting

Good infrastructure and smart classrooms

Assignments are very useful for understanding concepts

The teaching quality is excellent

Labs need more advanced datasets

Machine learning sessions are very useful

Good faculty and supportive teaching

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS

# 1. Create the dataset from the lab manual 
feedback_data = [
    "Teaching is excellent and very clear",
    "Labs are well organized and very helpful",
    "The faculty explains machine learning clearly",
    "Python lab sessions are very interesting",
    "Good infrastructure and smart classrooms",
    "Assignments are very useful for understanding concepts",
    "The teaching quality is excellent",
    "Labs need more advanced datasets",
    "Machine learning sessions are very useful",
    "Good faculty and supportive teaching"
]

# Combine all feedback into one string
text = " ".join(feedback_data)

# 2. Generate the Word Cloud
wordcloud = WordCloud(
    width=800, 
    height=400, 
    background_color='white',
    stopwords=STOPWORDS # Practice Question: Automatically removes common words
).generate(text)

# 3. Display the visualization
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Student Feedback Word Cloud")
plt.show()
```

# Lab Task 6: Create heatmap for crime data.

```python
import folium
from folium.plugins import HeatMap
import pandas as pd
import numpy as np

# 1. Initialize map centered on a city (e.g., Hyderabad) [cite: 54, 71]
m = folium.Map(location=[17.3850, 78.4867], zoom_start=12)

# 2. Create a dummy crime dataset [cite: 56, 57]
# We generate random coordinates around the city center
data = {
    'Lat': [17.3850, 17.3900, 17.3950, 17.4000, 17.3700, 17.3600, 17.4100, 17.3855, 17.3845],
    'Lon': [78.4867, 78.4800, 78.5000, 78.4700, 78.4900, 78.4800, 78.5200, 78.4860, 78.4870],
    'Intensity': [10, 5, 8, 2, 7, 3, 9, 15, 12] # Represents number of crimes
}
df = pd.DataFrame(data)

# 3. Prepare data for the HeatMap
# HeatMap expects a list of [lat, lon, weight] 
heat_data = [[row['Lat'], row['Lon'], row['Intensity']] for index, row in df.iterrows()]

# 4. Add HeatMap layer to the map
HeatMap(heat_data, radius=15, blur=10, max_zoom=1).add_to(m)

# 5. Save and display
m.save("crime_heatmap.html")
m
```

# Lab Task 7: Visualize COVID cases by state (dummy dataset).

```python
import folium
import pandas as pd

# 1. Initialize map centered on India [cite: 53]
m = folium.Map(location=[20.5937, 78.9629], zoom_start=5)

# 2. Create a Dummy Dataset for COVID cases [cite: 57]
covid_data = {
    'State': ['Andhra Pradesh', 'Telangana', 'Maharashtra', 'Karnataka', 'Tamil Nadu', 'Delhi', 'Uttar Pradesh'],
    'Cases': [5000, 3500, 12000, 8000, 9500, 11000, 7000]
}
df = pd.DataFrame(covid_data)

# 3. Coordinates for the centers of these states
state_coords = {
    'Andhra Pradesh': [15.9129, 79.7400],
    'Telangana': [17.3850, 78.4867],
    'Maharashtra': [19.7507, 75.7139],
    'Karnataka': [15.3173, 75.7139],
    'Tamil Nadu': [11.1271, 78.6569],
    'Delhi': [28.7041, 77.1025],
    'Uttar Pradesh': [26.8467, 80.9462]
}

# 4. Add Circle Markers based on Case Load [cite: 58]
for state, coord in state_coords.items():
    # Get the case count for the current state
    case_count = df.loc[df['State'] == state, 'Cases'].values[0].item()
    
    folium.Circle(
        location=coord,
        radius=case_count * 10, # Scaling radius to make it visible on the map [cite: 58, 83]
        popup=f"{state}: {case_count} cases",
        color='crimson',        # Red border
        fill=True,
        fill_color='crimson',   # Red fill
        fill_opacity=0.4
    ).add_to(m)

# 5. Save the map to an HTML file
# m.save("covid_cases_india.html")
m
```

# Lab Task 8: Add circle markers based on population.

```python
import folium
import pandas as pd

# 1. Initialize map centered on India
m = folium.Map(location=[20.5937, 78.9629], zoom_start=5)

# 2. Dataset from Lab Task 10 
data = {
    'City': ['Hyderabad', 'Mumbai', 'Delhi', 'Bangalore'],
    'Population': [10, 20, 18, 12], # Population in millions
    'Latitude': [17.38, 19.07, 28.70, 13.08],
    'Longitude': [78.48, 72.87, 77.10, 80.27]
}
df = pd.DataFrame(data)

# 3. Add Circle Markers based on Population 
for index, row in df.iterrows():
    # Convert numpy types to native Python types to avoid JSON errors
    pop_value = row['Population']
    
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=pop_value * 1.5, # Scaling factor for visibility
        popup=f"{row['City']}: {pop_value} Million",
        color='blue',
        fill=True,
        fill_color='blue',
        fill_opacity=0.6
    ).add_to(m)

# 4. Save and display
m.save("population_markers.html")
m
# If using Jupyter, type 'm' to see the map.
```

# Lab Task 9:Multi-Level Waffle Chart (Hierarchical Data)

- Create nested waffle chart
- Encode:
  - Color → Gender
  - Blocks → Total count
- Normalize to 100 squares

```python
import matplotlib.pyplot as plt
from pywaffle import Waffle
import pandas as pd

data = {
'Department': ['CSE','CSE','ECE','ECE','MECH','MECH'],
'Gender': ['Male','Female','Male','Female','Male','Female'],
'Count': [80,40,50,40,45,15]
}
df = pd.DataFrame(data)

# We create a dictionary of plots, one for each department
fig = plt.figure(
    FigureClass=Waffle,
    plots={
        311: { # 3 rows, 1 column, plot 1
            'values': df[df['Department'] == 'CSE'].set_index('Gender')['Count'].to_dict(),
            'title': {'label': 'Department: CSE', 'loc': 'left'}
        },
        312: { # 3 rows, 1 column, plot 2
            'values': df[df['Department'] == 'ECE'].set_index('Gender')['Count'].to_dict(),
            'title': {'label': 'Department: ECE', 'loc': 'left'}
        },
        313: { # 3 rows, 1 column, plot 3
            'values': df[df['Department'] == 'MECH'].set_index('Gender')['Count'].to_dict(),
            'title': {'label': 'Department: MECH', 'loc': 'left'}
        },
    },
    rows=5, 
    columns=20, # Normalizing to 100 squares 
    colors=["#3498db", "#e74c3c"], # Blue for Male, Red for Female [cite: 63]
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    figsize=(12, 8)
)

plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

from pywaffle import Waffle
import pandas as pd

from wordcloud import WordCloud, STOPWORDS

# Dataset from Lab Task 10 [cite: 67-73]
data = {
    'City': ['Hyderabad', 'Mumbai', 'Delhi', 'Bangalore'],
    'Population': [10, 20, 18, 12]
}
df = pd.DataFrame(data)

# 1. Waffle Chart
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    columns=20, # Normalizing to 100 squares
    values=df.set_index('City')['Population'].to_dict(),
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    title={'label': 'City-wise Population Share', 'loc': 'left'},
    figsize=(10, 5)
)
plt.show()

# 2. Word Cloud

city_descriptions = """
Hyderabad is known for its rich history, pearls, and biryani. 
Mumbai is the financial capital of India, a bustling coastal city. 
Delhi is the historical heart of India with famous monuments and political power. 
Bangalore is the Silicon Valley of India, famous for its IT industry and pleasant weather.
"""

wordcloud = WordCloud(width=800, height=400, background_color='white', stopwords=STOPWORDS).generate(city_descriptions)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("City Descriptions Word Cloud")
plt.show()

# 3. Map Visualization

import folium
from folium.plugins import HeatMap

# Full Dataset from Lab Task 10 [cite: 67-73]
city_data = {
    'City': ['Hyderabad', 'Mumbai', 'Delhi', 'Bangalore'],
    'Population': [10, 20, 18, 12],
    'Pollution': [70, 90, 85, 60],
    'Latitude': [17.38, 19.07, 28.70, 13.08],
    'Longitude': [78.48, 72.87, 77.10, 80.27]
}
df_map = pd.DataFrame(city_data)

# 1. Initialize map centered on India
m = folium.Map(location=[20.5937, 78.9629], zoom_start=5)

# 2. Add Heatmap based on Pollution [cite: 82]
heat_data = [[row['Latitude'], row['Longitude'], row['Pollution']] for index, row in df_map.iterrows()]
HeatMap(heat_data, radius=25, blur=15).add_to(m)

# 3. Add Circle Markers based on Population [cite: 80, 83]
for index, row in df_map.iterrows():
    pop_size = row['Population']
    poll_val = row['Pollution']
    
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=pop_size * 1.2, # Marker size -> Population [cite: 83]
        popup=f"{row['City']}<br>Population: {pop_size}M<br>Pollution Index: {poll_val}",
        color='black',
        fill=True,
        fill_color='blue',
        fill_opacity=0.6
    ).add_to(m)

# 4. Save and display
# m.save("city_analysis_map.html")
m
# Type 'm' in Jupyter to view
```

# Lab Task 11: Year-wise Market Share Comparison

```python
import pandas as pd
import matplotlib.pyplot as plt
from pywaffle import Waffle

# 1. Define the dataset from Lab Task 11 [cite: 85-89]
data = {
    'Year': [2022, 2022, 2022, 2023, 2023, 2023],
    'Company': ['A', 'B', 'C', 'A', 'B', 'C'],
    'Share': [50, 30, 20, 40, 35, 25]
}
df = pd.DataFrame(data)

# 2. Extract shares for each year as dictionaries
# Using .item() or casting to int avoids potential JSON/rendering errors with int64
share_2022 = df[df['Year'] == 2022].set_index('Company')['Share'].to_dict()
share_2023 = df[df['Year'] == 2023].set_index('Company')['Share'].to_dict()

# 3. Create separate waffle charts for each year in one figure 
fig = plt.figure(
    FigureClass=Waffle,
    plots={
        211: {  # 2 rows, 1 column, first plot
            'values': share_2022,
            'title': {'label': 'Market Share 2022', 'loc': 'left'}
        },
        212: {  # 2 rows, 1 column, second plot
            'values': share_2023,
            'title': {'label': 'Market Share 2023', 'loc': 'left'}
        },
    },
    rows=5,
    columns=20,  # 100 blocks total to represent percentages [cite: 95]
    colors=["#27ae60", "#2980b9", "#8e44ad"], # Custom colors for A, B, and C
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    figsize=(10, 8)
)

plt.tight_layout()
plt.show()
```
