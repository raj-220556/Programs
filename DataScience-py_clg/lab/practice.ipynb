```python
# OOPS Concept

class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
    def calculate(self):
        grade = ''
        per = sum(self.marks)/len(self.marks)
        if per > 90:
            grade = "A"
        elif per > 80:
            grade = "B"
        else:
            grade = "C"
        print(f"{self.name} obtained {grade} grade with {per}% percentage ")
s = Student("raj", 20, [90,80,97,93])
s.calculate()

# Multiple Inheritance

        
```
