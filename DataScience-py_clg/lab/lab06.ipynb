# One-Way ANOVA Problem

A teacher wants to check whether three different teaching methods produce different student scores. 

### Student Scores Data

| Method A | Method B | Method C |
| :---: | :---: | :---: |
| 78 | 85 | 92 |
| 82 | 88 | 95 |
| 85 | 90 | 89 |
| 80 | 87 | 94 |

---

### Tasks:
1. **State** null and alternative hypotheses.
2. **Calculate** group means.
3. **Calculate** Total Sum of Squares (SST).
4. **Calculate** Sum of Squares Between (SSB).
5. **Calculate** Sum of Squares Within (SSW).
6. **Construct** ANOVA table.
7. **Test** at $\alpha = 0.05$.
8. **Write** conclusion.

```python
import numpy as np
from scipy import stats
import pandas as pd

# 1. State Null and Alternative Hypotheses
null_hypothesis = "H0: There is no significant difference between the means of the three teaching methods (mu_A = mu_B = mu_C)."
alt_hypothesis = "Ha: At least one teaching method mean is significantly different from the others."

print("--- Statistical Hypotheses ---")
print(null_hypothesis)
print(alt_hypothesis)
print("-" * 30)

# 2. Calculate group means.
data = {
    'Method_A': [78, 82, 85, 80],
    'Method_B': [85, 88, 90, 87],
    'Method_C': [92, 95, 89, 94]
}

df = pd.DataFrame(data)

group_means = df.mean()
print("--- Group Means ---")
print(group_means)

# grand_mean = df.values.mean()
# print("\n--- Grand Mean ---")
# print(f"Overall average of all students: {grand_mean:.2f}")


# 3. Calculate Total Sum of Squares (SST).

# Convert to a flat NumPy array
all_scores = df.values.flatten()

# # Convert to a single Pandas Series
# all_scores_series = df.stack()

# # If you need it as a simple list:
# all_scores_list = df.stack().tolist()

# Calculate Total Sum of Squares (SST)
# (Subtract grand mean from each score, square the result, and sum them up)
grand_mean = df.values.mean()
sst = np.sum((all_scores - grand_mean)**2)

print(f"Grand Mean: {grand_mean:.2f}")
print(f"Total Sum of Squares (SST): {sst:.2f}")


# 4. Calculate Sum of Squares Between (SSB).

# Calculate SSB
# We take each column mean, subtract the grand mean, square it, 
# multiply by the number of elements in that column, and sum them up.
ssb = (df.count() * (df.mean() - grand_mean)**2).sum()
print(f"Sum of Squares Between (SSB): {ssb:.2f}")


# 5. Calculate Sum of Squares Within (SSW).

# Calculate SSW
# We subtract each column's mean from its respective values, 
# square those differences, and then sum everything.
group_means = df.mean()
ssw = ((df - group_means)**2).values.sum()
print(f"Sum of Squares Within (SSW): {ssw:.2f}")


# 6. Construct ANOVA table.

# Basic Variables
N = df.size               # Total number of observations (12)
k = len(df.columns)       # Number of groups (3)
grand_mean = df.values.mean()

# Calculate Sum of Squares (SS)
ssb = (df.count() * (df.mean() - grand_mean)**2).sum()
ssw = ((df - df.mean())**2).values.sum()
sst = ssb + ssw

# Degrees of Freedom (df)
df_between = k - 1
df_within = N - k
df_total = N - 1

# Mean Squares (MS)
msb = ssb / df_between
msw = ssw / df_within

# F-Statistic
f_stat = msb / msw

# Construct and Display the Table
anova_table = pd.DataFrame({
    'Source of Variation': ['Between Groups (Treatment)', 'Within Groups (Error)', 'Total'],
    'Sum of Squares (SS)': [ssb, ssw, sst],
    'Degrees of Freedom (df)': [df_between, df_within, df_total],
    'Mean Square (MS)': [msb, msw, np.nan],
    'F-Statistic': [f_stat, np.nan, np.nan]
})

print("--- ONE-WAY ANOVA TABLE ---")
print(anova_table.to_string(index=False))


# 7. Test at α = 0.05.

# Perform ANOVA to get F-statistic and P-value
f_stat, p_value = stats.f_oneway(df['Method_A'], df['Method_B'], df['Method_C'])

alpha = 0.05
df_between = len(df.columns) - 1
df_within = df.size - len(df.columns)

# stats.f.ppf returns the critical value for the F-distribution
f_critical = stats.f.ppf(1 - alpha, df_between, df_within)

print(f"--- Hypothesis Testing (alpha = {alpha}) ---")
print(f"Degrees of Freedom: df1={df_between}, df2={df_within}")
print(f"Calculated F-Statistic: {f_stat:.4f}")
print(f"Critical F-Value: {f_critical:.4f}")
print(f"P-Value: {p_value:.6f}")

print("-" * 40)

# 8. Write conclusion.

if f_stat > f_critical:
    conclusion = (
        "Decision: Reject the Null Hypothesis (H0).\n"
        "Conclusion: Since the calculated F-statistic is greater than the critical value "
        "(and the P-value < 0.05), there is significant evidence to conclude that the "
        "three teaching methods produce different student scores."
    )
else:
    conclusion = (
        "Decision: Fail to Reject the Null Hypothesis (H0).\n"
        "Conclusion: There is no significant evidence to suggest that the "
        "teaching methods produce different student scores."
    )

print(conclusion)
```

# 2. Two-Way ANOVA Question

A researcher is studying the effect of two different factors on crop yield.

### Factors and Levels:
* **Factor 1: Fertilizer Type** (Levels: A, B)
* **Factor 2: Water Level** (Levels: Low, High)

---

### Crop Yield Data Table

| Fertilizer | Low Water | High Water |
| :--- | :--- | :--- |
| **Fertilizer A** | 30, 32, 31 | 40, 42, 41 |
| **Fertilizer B** | 28, 29, 27 | 38, 37, 39 |

---

### Tasks:
1.  **Identify** the independent and dependent factors.
2.  **Test** the main effects of Fertilizer Type.
3.  **Test** the main effects of Water Level.
4.  **Test** for an **interaction effect** between Fertilizer and Water.
5.  **Write** a final interpretation of the results at $\alpha = 0.05$.

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

data = {
    'Fertilizer': ['A', 'A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'B', 'B'],
    'Water_Level': ['Low', 'Low', 'Low', 'High', 'High', 'High', 
                    'Low', 'Low', 'Low', 'High', 'High', 'High'],
    'Yield': [30, 32, 31, 40, 42, 41, 28, 29, 27, 38, 37, 39]
}

df = pd.DataFrame(data)

# 1. Identify Factors (Categorical Columns)
# We exclude the dependent variable (Yield) which is numerical
factors = df.select_dtypes(include=['object']).columns.tolist()

print(f"Detected Factors: {factors}")

# Identify Levels for each factor
for factor in factors:
    levels = df[factor].unique()
    print(f"Factor: {factor} | Levels: {levels} | Count: {len(levels)}")

#  Identify Dependent Variable
dependent_var = 'Yield'
print(f"Dependent Variable: {dependent_var}")


# 2. Test main effects.

# 'Yield ~ C(Fertilizer) + C(Water)' tells the model to look for Main Effects
# We include the interaction '*' to ensure the main effects are calculated correctly
model = ols('Yield ~ C(Fertilizer) * C(Water_Level)', data=df).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print("--- ANOVA Results for Main Effects ---")
print(anova_table.loc[['C(Fertilizer)', 'C(Water_Level)']])

# Interpretation Helper
for factor in ['C(Fertilizer)', 'C(Water_Level)']:
    p_val = anova_table.loc[factor, 'PR(>F)']
    status = "Significant" if p_val < 0.05 else "Not Significant"
    print(f"\n{factor}: p-value = {p_val:.4f} ({status})")


# 3. Test interaction effect.

interaction_row = 'C(Fertilizer):C(Water_Level)'
p_val = anova_table.loc[interaction_row, 'PR(>F)']

print("--- Interaction Effect Analysis ---")
print(anova_table.loc[[interaction_row]])

# Interpretation
if p_val < 0.05:
    print(f"\nResult: Significant Interaction (p = {p_val:.4f})")
    print("Interpretation: The choice of Fertilizer affects Yield differently depending on the Water Level.")
else:
    print(f"\nResult: No Significant Interaction (p = {p_val:.4f})")
    print("Interpretation: The Fertilizer effect is consistent across both Water Levels.")


# def interpret_anova(table):
#     print("=== RESEARCH INTERPRETATION ===\n")
    
#     # Extract p-values
#     p_fert = table.loc['C(Fertilizer)', 'PR(>F)']
#     p_water = table.loc['C(Water_Level)', 'PR(>F)']
#     p_inter = table.loc['C(Fertilizer):C(Water_Level)', 'PR(>F)']
    
#     # 1. Interpret Main Effect: Fertilizer
#     if p_fert < 0.05:
#         print(f"1. Fertilizer Effect: SIGNIFICANT (p = {p_fert:.4f}).")
#         print("   The type of fertilizer used has a real impact on crop yield.")
#     else:
#         print(f"1. Fertilizer Effect: NOT SIGNIFICANT (p = {p_fert:.4f}).")
#         print("   There is no evidence that one fertilizer brand is better than the other.")

#     # 2. Interpret Main Effect: Water Level
#     if p_water < 0.05:
#         print(f"2. Water Level Effect: SIGNIFICANT (p = {p_water:.4f}).")
#         print("   Changes in water levels significantly alter the crop yield.")
#     else:
#         print(f"2. Water Level Effect: NOT SIGNIFICANT (p = {p_water:.4f}).")
#         print("   The amount of water provided did not statistically change the yield.")

#     # 3. Interpret Interaction Effect
#     print("-" * 30)
#     if p_inter < 0.05:
#         print(f"3. Interaction Effect: SIGNIFICANT (p = {p_inter:.4f}).")
#         print("   WARNING: The best fertilizer depends on the water level.")
#         print("   One fertilizer might thrive in Low water but fail in High water.")
#     else:
#         print(f"3. Interaction Effect: NOT SIGNIFICANT (p = {p_inter:.4f}).")
#         print("   The fertilizers behave consistently regardless of the water level.")
#         print("   You can choose the best fertilizer and best water level independently.")

# # Execute the interpretation
# interpret_anova(anova_table)
```
