```python
# 1.Write a program to create a class for student data (student ID, name, marks, and
# percentage). Implement a method to calculate the percentage and grade (A, B, C, D)
# based on the percentage. Take the values as input at runtime.


class Student:
    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks  # Expecting a list of marks
        self.percentage = 0.0
        self.grade = ""

    def calculate_results(self):
        # Calculate percentage (assuming each subject is out of 100)
        total_marks = sum(self.marks)
        self.percentage = total_marks / len(self.marks)
        
        # Determine grade based on percentage
        if self.percentage >= 85:
            self.grade = 'A'
        elif self.percentage >= 70:
            self.grade = 'B'
        elif self.percentage >= 50:
            self.grade = 'C'
        else:
            self.grade = 'D'

    def display_data(self):
        print("\n--- Student Report Card ---")
        print(f"ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Percentage: {self.percentage:.2f}%")
        print(f"Grade: {self.grade}")

# --- Runtime Input ---
s_id = input("Enter Student ID: ")
s_name = input("Enter Student Name: ")
# Taking multiple marks separated by space and converting them to integers
s_marks = list(map(int, input("Enter marks for 3 subjects (separated by space): ").split()))

# Create object and call methods
student1 = Student(s_id, s_name, s_marks)
student1.calculate_results()
student1.display_data()
```

```python
# 2.Write a program to demonstrate multiple inheritance where one class represents a
# person and another represents a teacher. The derived class should inherit both
# functionalities and display teacher and person details.


# Parent Class 1: Person
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

# Parent Class 2: Professional Details
class TeacherDetails:
    def __init__(self, subject, salary):
        self.subject = subject
        self.salary = salary

    def display_teacher(self):
        print(f"Subject: {self.subject}")
        print(f"Salary: ${self.salary}")

# Derived Class: Inherits from both Person and TeacherDetails
class ProfessionalTeacher(Person, TeacherDetails):
    def __init__(self, name, age, subject, salary):
        # Initialize both parent classes
        Person.__init__(self, name, age)
        TeacherDetails.__init__(self, subject, salary)

    def show_full_profile(self):
        print("--- Teacher Profile ---")
        self.display_person()
        self.display_teacher()

# --- Execution ---
name = input("Enter Name: ")
age = int(input("Enter Age: "))
subj = input("Enter Subject: ")
sal = float(input("Enter Salary: "))

# Creating the object of the derived class
t1 = ProfessionalTeacher(name, age, subj, sal)
t1.show_full_profile()
```

```python
# 3.Write a program to validate a phone number using regular expressions. Ensure the number
# follows the format: XXX-XXX-XXXX.

import re

phone_number = input("Enter a phone number (Format: XXX-XXX-XXXX): ")
pattern = r"^\d{3}-\d{3}-\d{4}$"
if re.match(pattern, phone_number):
    print(f"{phone_number} is a valid phone number.")
else:
    print(f"Invalid format. Please use 123-456-7890.")
```

```python
# 4.Write a program that accepts a list of student scores and performs the following tasks:
# a.​ Sort the list in ascending order.
# b.​ Identify and remove duplicate scores.
# c.​ Calculate the average score.

def process_scores(scores):
    # b. Identify and remove duplicate scores
    # Converting to a set automatically removes duplicates
    unique_scores = list(set(scores))
    
    # a. Sort the list in ascending order
    unique_scores.sort()
    
    # c. Calculate the average score
    average = sum(unique_scores) / len(unique_scores)
        
    return unique_scores, average

# --- Runtime Input ---
raw_input = input("Enter student scores separated by spaces: ")
# Convert input string into a list of integers
score_list = [int(x) for x in raw_input.split()]

# Processing
sorted_unique, avg = process_scores(score_list)

# Display Results
print("\n--- Results ---")
print(f"Original List: {score_list}")
print(f"Sorted (No Duplicates): {sorted_unique}")
print(f"Average of unique scores: {avg:.2f}")
```

```python
# 5.Write a program to simulate a simple library system. Implement classes for Book and Member.
# Allow members to borrow and return books, and track which books are currently borrowed.

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_borrowed = False

    def __str__(self):
        status = "Borrowed" if self.is_borrowed else "Available"
        return f"'{self.title}' by {self.author} [{status}]"

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book):
        if not book.is_borrowed:
            book.is_borrowed = True
            self.borrowed_books.append(book)
            print(f"{self.name} successfully borrowed '{book.title}'.")
        else:
            print(f"Sorry, '{book.title}' is already borrowed by someone else.")

    def return_book(self, book):
        if book in self.borrowed_books:
            book.is_borrowed = False
            self.borrowed_books.remove(book)
            print(f"{self.name} returned '{book.title}'.")
        else:
            print(f"{self.name} does not have this book.")

book1 = Book("1984", "shiva")
book2 = Book("The Lion", "raj")
member1 = Member("Alice", "M001")

# print(book1)
# print(book2)

member1.borrow_book(book1)
member1.borrow_book(book2)

member2 = Member("Bob", "M002")
member2.borrow_book(book1)

member1.return_book(book1)

print(f"\nBooks currently with {member1.name}: {[b.title for b in member1.borrowed_books]}")
```
