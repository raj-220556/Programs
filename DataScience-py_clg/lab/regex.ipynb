```python
# 1. In the regex created from r'(\d\d\d)-(\d\d\d-\d\d\d\d)',
# what does group 0 cover? Group 1? Group 2?

## Group 0 : (\d\d\d)-(\d\d\d-\d\d\d\d)  :- Entire match
## Group 1 : (\d\d\d)
## Group 2 : (\d\d\d-\d\d\d\d)
import re

pattern = re.compile(r'(\d\d\d)-(\d\d\d-\d\d\d\d)')
obj = pattern.search('My number is 415-555-4242.')

print(f"Group 0: {obj.group(0)}") # The whole match
print(f"Group 1: {obj.group(1)}") # The first ()
print(f"Group 2: {obj.group(2)}") # The second ()
```

```python
# 2. Find website URLs that begin with http:// or https://.

url_pattern = r"https?://([\w-]+\.)*[\w-]+\.(com|org|net|onion|app)(/[^\s<>]*)?"
url = "https://user.github.com/index.php?id=10&password=none"

if (re.match(url_pattern,url)):
    print("IT is a URL")
else:
    print("IT might be something else not a URL")
```

```python
# 3. Clean up dates in different date formats (such as 3/14/2015, 03-14-2015, and 2015/3/14) by
# replacing them with dates in a single, standard format(YYYY-MM-DD).


date_pattern = r"(\d{1,2})[\./-](\d{1,2})[\./-](\d{4})" # checks DD.MM.YYYY or MM.DD.YYYY
replacement = r"\3-\1-\2"
text = "2025/12/1, 3/14/2015 2015/3/14 03-14-2015 3.14.2015"

print(re.sub(date_pattern, replacement, text))
```

```python
# 4. How would you write a regex that matches a number with commas for every three digits? It
# must match the following:
# • '42'
# • '1,234'
# • '6,368,745'
# but not the following:
# • '12,34,567' (which has only two digits between the commas)
# • '1234' (which lacks commas)

import re

num_pattern = r"^\d{1,3}(,\d{3})*$"

test_numbers = ["42", "1,234", "6,368,745", "12,34,567", "1234", "1,00" ]

for num in test_numbers:
    if re.match(num_pattern, num):
        print(f"'{num}': MATCH")
    else:
        print(f"'{num}': NO MATCH")
```

```python
# 5. How would you write a regex that matches the full name of someone whose last name is
# Nakamoto? You can assume that the first name that comes before it will always be one word
# that begins with a capital letter.
# The regex must match the following:
# • 'Satoshi Nakamoto'
# • 'Alice Nakamoto'
# • 'RoboCopNakamoto'
# but not the following:
# • 'satoshiNakamoto' (where the first name is not capitalized)
# • 'Mr. Nakamoto' (where the preceding word has a nonletter character)
# • 'Nakamoto' (which has no first name)
# • 'Satoshi nakamoto' (where Nakamoto is not capitalized)

name_pattern = r"^([A-Z][a-zA-Z]* ?)Nakamoto$"
names = ['Satoshi Nakamoto', 'Alice Nakamoto', 'RoboCopNakamoto', 'satoshiNakamoto', 'Mr. Nakamoto', 'Nakamoto', 'Satoshi nakamoto']

for name in names:
    if re.match(name_pattern, name):
        print(f"{name} is Match")
    else:
        print(f"{name} Not a Match")

```

```python
# 6. How would you write a regex that matches a sentence where the first word is either Alice, Bob,
# or Carol; the second word is either eats, pets, or throws; the third word is apples, cats, or
# baseballs; and the sentence ends with a period? This regex should be case-insensitive. It must
# match the following:
# • 'Alice eats apples.'
# • 'Bob pets cats.'
# • 'Carol throws baseballs.'
# • 'Alice throws Apples.'
# • 'BOB EATS CATS.'
# but not the following:
# • 'RoboCop eats apples.'
# • 'ALICE THROWS FO
# • 'Carol eats 7 cats.'

sentense_pattern = r"^(Alice|Bob|Carol)\s(eats|pets|throws)\s(apples|cats|baseballs)\.$"
sentenses = [
    'Alice eats apples.',
    'Bob pets cats.',
    'Carol throws baseballs.',
    'Alice throws Apples.',
    'BOB EATS CATS.',
    'RoboCop eats apples.',
    'ALICE THROWS FO',
    'Carol eats 7 cats.'
]

for sent in sentenses:
    if re.match(sentense_pattern,sent,re.IGNORECASE):
        print(f"'{sent}' : Match")
    else:
        print(f"'{sent}' : Not a Match")
```

```python
# 7. Strong Password Detection Write a function that uses regular expressions to make sure the
# password string it is passed is strong. A strong password is defined as one that is at least eight
# characters long, contains both uppercase and lowercase characters, and has at least one digit.
# You may need to test the string against multiple regex patterns to validate its strength.

def isStrongPassword(str):
    if len(str) < 8:
        return False
    flag = 0
    if re.search(r"[A-Z]", str): 
        flag+=1
    else: return False
    if re.search(r"[a-z]", str): 
        flag+=1
    else: return False
    if re.search(r"\d", str):
        flag+=1
    else: return False

    return True

print(isStrongPassword("Aqwert19"))  
print(isStrongPassword("Aqwertqw"))

```
