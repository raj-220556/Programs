### 7.Question: write a python program to input lines of text from the user until enter is pressed, count the total number of characters in the text including whitespaces, total number of alphabets, digits and special symbols and total number of words in the given text.

```python
s = input()
c,d,sp = 0,0,0
for i in s:
    if i.isdigit():
        d += 1
    elif i.isalpha():
        c += 1
    else:
        sp += 1
print("Total no.of Charcters:",len(s))
print("Total no.of Alphabets:",c)
print("Total no.of digits:",d)
print("Total no.of Special Charcters:",sp)
print("Total no.of Words:",len(s.split()))
```

### 2.Question: Write a python program to convert a number entered by user into a corresponding number in words


```python
n = int(input("Enter Number:"))
#crores,lakhs,thousand,hundred
#Program
word = { 0:"",1:'one', 2:'two', 3:'three', 4:'four', 5:'five', 6:'six', 7:'seven', 8:'eight',
         9:'nine', 10:'ten', 11:'eleven', 12:'twelve', 13:'thirteen', 14:'forteen',
         15:'fifteen', 16:'sixteen', 17:'seventeen', 18:'eighteen', 19:'nineteen',
         20:'twenty', 30:'thirty', 40:'forty', 50:'fifty', 60:'sixty', 70:'seventy',
         80:'eighty', 90:'ninety'
        }
output = []
num = n
def spellTwo(n,tens):
    if(word.get(tens,-1) == -1):
            t = tens - n%10
            output.insert(0,word.get(n%10))
            output.insert(0,word.get(t))
    else:
        output.insert(0,word.get(tens))
while(n > 0):
    tens = n%100
    if(tens > 0):
        spellTwo(n,tens)
        output.insert(0,'and')
    n //= 100
    hun = n%10
    if(hun > 0):
        output.insert(0,'hundred')
        output.insert(0,word.get(hun))
    n //= 10
    thou = n%100
    if(thou > 0):
        output.insert(0,'thousand')
        spellTwo(n,thou)
    n //= 100
    lakh = n%100
    if(lakh > 0):
        output.insert(0,'lakh')
        spellTwo(n,lakh)
    n //= 100
    if(n > 0):
        output.insert(0,'crore')
# print(output)
print(num)
" ".join(output)
```

### 2.Question: To print prime no;s b/w 2 and 50

```python
import math
#progaram:
def prime(n):
    if n == 2: return True
    for i in range(2,math.ceil(math.sqrt(n))+1):
        if n%i == 0:
            return False
    return True
for i in range(2,50):
    if(prime(i)):
        print(i,end=", ")
print("\b\b.")
```

```python

#create a class called students with attributes name , id, branch , methods precentage, grade. display the above details id,name,brance,percentance,grade
#for at least two students
class student:
    def __init__(self,name, id , branch, marks):
        self.name = name
        self.id = id
        self.branch = branch
        self.marks = marks
    def percentage(self):
        return (self.marks/600)*100
    def grade(self):
        if(self.marks > 550):
            return "A"
        elif(self.marks > 500):
            return "B"
        elif(self.marks > 450):
            return "C"
        elif(self.marks > 400):
            return "D"
        else:
            return "F"
std1 = student("Jyosh",210903,"CSE",567)
print(std1.id, std1.name, std1.branch, std1.percentage(), std1.grade())

std2 = student("suji",210015,"CSE",499)
print(std2.id, std2.name,  std2.branch, std2.percentage(), std2.grade())
```
